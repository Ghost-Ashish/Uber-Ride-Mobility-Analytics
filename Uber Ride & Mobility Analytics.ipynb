```python
# Clean all text columns
text_columns = df_clean.select_dtypes(include="object").columns

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()

print("Text columns cleaned successfully.")
```

```python
# Convert Date
df_clean["date"] = pd.to_datetime(
    df_clean["date"],
    errors="coerce"
)

# Convert Time
df_clean["time"] = pd.to_datetime(
    df_clean["time"],
    format="%H:%M:%S",
    errors="coerce"
).dt.time

print(df_clean[["date", "time"]].dtypes)
display(df_clean[["date", "time"]].head())
```

```python
# Extract date-related features
df_clean["year"] = df_clean["date"].dt.year
df_clean["month"] = df_clean["date"].dt.month
df_clean["month_name"] = df_clean["date"].dt.month_name()
df_clean["day"] = df_clean["date"].dt.day
df_clean["day_name"] = df_clean["date"].dt.day_name()
df_clean["weekday"] = df_clean["date"].dt.dayofweek

# Extract hour from original time strings
df_clean["hour"] = pd.to_datetime(
    df_raw["Time"],
    format="%H:%M:%S",
    errors="coerce"
).dt.hour

display(
    df_clean[
        ["date", "year", "month_name", "day_name", "hour"]
    ].head()
)
```

```python
def get_time_period(hour):
    if pd.isna(hour):
        return "Unknown"
    elif 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df_clean["time_period"] = df_clean["hour"].apply(get_time_period)
```
